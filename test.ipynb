```python
## Real example:

## student management system

class Student:

    ## constructor ## method
    def __init__(self, name, age, roll_no):
        self.name = name
        self.age = age
        self.roll_no = roll_no

    def display(self):
        print("Name:", self.name)
        print("Age:", self.age)
        print("Roll No:", self.roll_no)

```

```python
s1 = Student("Sachin", 25, 101)
print(s1)
```

```python
s1.display()
```

```python
class Student:

    ## constructor ## method
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    def calculate_grade(self):
        if self.marks >= 90:
            return "A-Grade"
        elif self.marks >= 75:
            return "B-Grade"
        elif self.marks >= 60:
            return "C-Grade"
        else:
            return "D-Grade"

```

```python

s1 = Student("Rahul", 82)

print("Grade:", s1.calculate_grade())
```

## Bank Management/ ATM Code

```python
class BankAccount:
    ## arguments: When we pass values to parameters, those values are called arguments
    def __init__(self, name, balance):  ## constructor , method
        self.name = name
        self.balance = balance

    def deposit(self, amount):   #method because they are within class
            self.balance += amount
            print(f"Deposited {amount}. New balance: {self.balance}")

    def withdraw(self, amount):
            if amount <= self.balance:
                self.balance -= amount
                print(f"Withdrew {amount}. New balance: {self.balance}")
            else:
                print("Insufficient balance")

    def display_balance(self):
            print("Balance:", self.balance)
    


```

```python
sachin_account = BankAccount("Sachin", 10000)  

    

  
```

```python
sachin_account.deposit(5000)
```

```python
sachin_account.withdraw(3000)
```

```python

sachin_account.display_balance()
```

```python
class Calculator:
    def add(self, a, b):
        print(a + b)

    def subtract(self, a, b):
        print(a - b)

    def multiply(self, a, b):
        print(a * b)

    def divide(self, a, b):
        print(a / b)


```

```python
calc = Calculator()

```

```python
calc.add(10, 5)
calc.subtract(10, 5)
calc.multiply(10, 5)
calc.divide(10, 5)
```

## Lambda Functions


## Add two numbers


```python
add = lambda a, b: a + b

print(add(10, 20))
```

## Check whether a number is even

```python
is_even = lambda x: x % 2 == 0

print(is_even(10))
print(is_even(7))
```

## Sort a list based on a specific value

```python
students = [
    ("Sachin", 85),
    ("Rahul", 72),
    ("Amit", 91),
    ("Priya", 78)
]

students.sort(key=lambda x: x[1])

print(students)
```

## lambda x: x[1]

### means use the second element of each tuple for sorting.

## Filter numbers using filter()

```python
numbers = [10, 25, 55, 70, 40, 90, 30]

result = list(filter(lambda x: x > 50, numbers))

print(result)
```
